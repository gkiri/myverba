```python
def semantic_search(query: str, limit: int = 10):
    """Perform semantic search on PYQS questions"""
    result = (
        client.query
        .get("PYQS", ["question", "options", "answer_key", "description"])
        .with_near_text({"concepts": [query]})
        .with_limit(limit)
        .with_additional(["distance"])
        .do()
    )

    if not result['data']['Get']['PYQS']:
        return []

    return [
        {
            "question": item["question"],
            "answer": item["answer_key"],
            "confidence": 1 - item["_additional"]["distance"],  # Convert to similarity score
            "distance": item["_additional"]["distance"],  # Raw distance value
            "explanation": item["description"]
        }
        for item in result['data']['Get']['PYQS']
    ]

# Example usage
results = semantic_search(subtopic3)
print(f"Found {len(results)} relevant questions:")
for idx, result in enumerate(results, 1):
    print(f"\nQuestion {idx}:")
    print(f"Score: {result['distance']:.4f} (Distance)")
    print(f"Confidence: {result['confidence']:.2%}")
    print(f"Q: {result['question']}")
    print(f"Explanation: {result['explanation']}")
```

#### BM25 search

```python
def bm25_search(query: str, limit: int = 10):
    """Perform BM25 search on PYQS questions"""
    result = (
        client.query
        .get("PYQS", ["question", "options", "answer_key", "description"])
        .with_bm25(
            query=query,
            properties=["question", "description"]  # Search these text fields
        )
        .with_limit(limit)
        .with_additional(["score"])  # BM25 relevance score
        .do()
    )

    if not result['data']['Get']['PYQS']:
        return []

    return [
        {
            "question": item["question"],
            "answer": item["answer_key"],
            "score": item["_additional"]["score"],
            "explanation": item["description"]
        }
        for item in result['data']['Get']['PYQS']
    ]

# Example usage
results = bm25_search(subtopic3)
print(f"Found {len(results)} relevant questions:")
for idx, result in enumerate(results, 1):
    print(f"\nQuestion {idx}:")
    print(f"BM25 Score: {float(result['score']):.4f}")  # Convert string to float first
    print(f"Q: {result['question']}")
    print(f"A: {result['answer']}\nExplanation: {result['explanation']}")
```

```python
# key points
# 1. Alpha Tuning: With alpha=0.7, results are weighted 70% toward vector search (semantic matches) and 30% toward BM25 (keywords).
# 2. Weaviate v3 client: doesnt support distance and explain score
# 3. Focus on the hybrid_score for ranking. If you need vector distances, consider migrating to Weaviate v4.
# 4. Trust the Hybrid Score: The hybrid_score (e.g., 0.7000) is the true relevance metric combining BM25 and vector search. Higher = better match.
# 5. Upgrade to Weaviate v4: To access distance, explainScore, and other hybrid search improvements.
# 6. Weaviate v3 Limitation: Hybrid search in v3 does not return distance by default. Your code sets a fallback of 1.0 (the worst possible cosine distance) when distance is missing.
# 7. Action: Remove the distance field from your code if you’re using v3, as it is not meaningful here.
# 8. The output shows all vector_distance values at 1.0, which indicates that Weaviate v3 is not returning the distance field for hybrid search results. This is a known limitation in older Weaviate versions. Here's what you need to do:

```

```python
def hybrid_search(query: str, limit: int = 5, alpha: float = 0.5):
    """Perform hybrid BM25 + vector search on PYQS questions"""
    result = (
        client.query
        .get("PYQS", ["question", "options", "answer_key", "description"])
        .with_hybrid(
            query=query,
            alpha=alpha,
            properties=["question", "description"]
        )
        .with_limit(limit)
        .with_additional(["score"])  # Only request hybrid `score`
        .do()
    )

    if not result['data']['Get']['PYQS']:
        return []

    return [
        {
            "question": item["question"],
            "answer": item["answer_key"],
            "hybrid_score": float(item["_additional"].get("score", 0.0)),
            "explanation": item["description"]
        }
        for item in result['data']['Get']['PYQS']
    ]


# Example usage
results = hybrid_search(subtopic3, limit=50, alpha=0.7)
sorted_results = sorted(results, key=lambda x: x["hybrid_score"], reverse=True)

print(f"\nFound {len(sorted_results)} relevant questions:\n")
for idx, result in enumerate(sorted_results, 1):
    print(f"╔{'═'*80}╗")
    print(f"║ Rank {idx}: Score = {result['hybrid_score']:.4f}")
    print(f"╠{'─'*80}╣")
    print(f"║ Question: {result['question']}")
    if result['explanation']:
        print(f"╠{'─'*80}╣")
        print(f"║ Explanation: {result['explanation'][:150]}...")  # Truncate long text
    print(f"╚{'═'*80}╝\n")
```

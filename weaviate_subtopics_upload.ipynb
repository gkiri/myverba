```python
import json
import re

def parse_custom_json(filename):
    try:
        with open(filename, 'r', encoding='utf-8') as file:
            content = file.read()

        # Extract file_name
        file_name_match = re.search(r'"file_name":\s*"([^"]*)"', content)
        file_name = file_name_match.group(1) if file_name_match else "Unknown"

        # Updated chapter pattern with non-greedy matching for subtopics array
        chapter_pattern = re.compile(
            r'\{([^{}]*?"chapter_number":[^{}]*?'
            r'"chapter_name":[^{}]*?'
            r'"subtopics_name":\s*\[[^\]]*?\]'  # Match subtopics_name array
            r'[^{}]*?"pagenumber_start_end":[^{}]*?'
            r'"ch_id":[^{}]*?'
            r'"chapter_content":\s*"(.*?)"'     # Match chapter_content
            r'[^{}]*?"subtopics":\s*\[(.*?)\]'  # Match subtopics array with objects
            r'\s*?\})',
            re.DOTALL
        )
        
        chapters = chapter_pattern.findall(content)

        # Process each chapter
        processed_chapters = []
        for chapter in chapters:
            chapter_dict = {}
            chapter_content = chapter[1]  # Second group is chapter_content
            subtopics_content = chapter[2]  # Third group is subtopics array content
            chapter_metadata = chapter[0] + chapter_content  # Combine for easier regex

            # Process regular fields
            for field in ["chapter_number", "chapter_name", "pagenumber_start_end", "ch_id"]:
                match = re.search(rf'"{field}":\s*"([^"]*)"', chapter_metadata)
                if match:
                    chapter_dict[field] = match.group(1)

            # Process subtopics_name array
            subtopics_name_match = re.search(r'"subtopics_name":\s*(\[[^\]]*\])', chapter_metadata)
            if subtopics_name_match:
                try:
                    subtopics_name = json.loads(subtopics_name_match.group(1))
                    chapter_dict["subtopics_name"] = subtopics_name
                except:
                    chapter_dict["subtopics_name"] = []

            # Process chapter_content
            chapter_dict["chapter_content"] = chapter_content.strip()

            # Process subtopics array with objects - Updated pattern
            subtopics = []
            subtopic_pattern = re.compile(
                r'\{\s*"subtopic_id":\s*"([^"]*)"'
                r'\s*,\s*"subtopic_name":\s*"([^"]*)"'
                r'\s*,\s*"order":\s*(\d+)'
                r'\s*,\s*"(?:content|subtopic_content)":\s*"(.*?)"'  # Handle both content and subtopic_content
                r'\s*\}',
                re.DOTALL
            )
            
            # Find all matches in the subtopics content
            for match in subtopic_pattern.finditer(subtopics_content):
                subtopic = {
                    "subtopic_id": match.group(1),
                    "subtopic_name": match.group(2),
                    "order": int(match.group(3)),
                    "subtopic_content": match.group(4).strip()
                }
                subtopics.append(subtopic)
            
            chapter_dict["subtopics"] = subtopics
            processed_chapters.append(chapter_dict)

        # Construct the final JSON structure
        result = {
            "file_name": file_name,
            "metadata": processed_chapters
        }

        # Debug information
        print(f"Found {len(processed_chapters)} chapters")
        for chapter in processed_chapters:
            print(f"Chapter {chapter.get('chapter_number', 'Unknown')}: {len(chapter.get('subtopics', []))} subtopics")

        return result

    except Exception as e:
        print(f"An error occurred while parsing the file: {str(e)}")
        return None


# Usage example:
filename = r"d:\UPSC-AI\Data\AI_Mentor\GS-1\json-gemini\subtopic-design\India_Ancient_Past_2017_sample.json"
#filename = r"d:\UPSC-AI\Data\AI_Mentor\GS-1\json-gemini\subtopic-design\Modern_India_Bipin_Chandra_sample.json"
result = parse_custom_json(filename)

if result:
    print(json.dumps(result, indent=2))
else:
    print("Failed to parse the file.")
```

```python
# Configure a batch process
with client.batch as batch:
    batch.batch_size=5
    for i, json_node in enumerate(result['metadata']):
        chapter_properties = {
            "chapter_name": json_node['chapter_name'],
            "chapter_number": 'NA',#json_node['chapter_number'],
            "subtopics_name": json_node['subtopics_name'],
            "pagenumber_start_end": json_node['pagenumber_start_end'] ,
            "ch_id": json_node['ch_id'],
            "chapter_content": json_node['chapter_content'],
        }

        # Create chapter object and get its ID
        chapter_uuid = client.batch.add_data_object(
            data_object=chapter_properties,
            class_name="VERBA_Syllabus_Chapters"
        )
        
        # Create subtopics for this chapter
        for subtopic in json_node.get('subtopics', []):
            subtopic_properties = {
                "subtopic_id": subtopic['subtopic_id'],
                "subtopic_name": subtopic['subtopic_name'],
                "order": subtopic['order'],
                "subtopic_content": subtopic.get('subtopic_content'),
            }
            
            # Create subtopic and link it to the chapter
            client.batch.add_data_object(
                data_object=subtopic_properties,
                class_name="VERBA_Syllabus_Subtopics"
            )
```

## Read or load json

```python
filename = r"d:\UPSC-AI\Data\mock prelims\data\qa-4701.json"
```

```python
import json

with open(filename, 'r', encoding='utf-8') as file:
    data = json.load(file)
```

```python
data
```

```python
# Copy question_number to global_questionID for each item
for item in data:
    item['global_questionID'] = item['question_number']

# Verify the change (optional)
print(data[0]['question_number'], data[0]['global_questionID'])
```

```python
data
```

```python
filename_out = r"d:\UPSC-AI\Data\mock prelims\data\qa-4701_out.json"
```

```python
# Save the updated data back to the file
with open(filename_out, 'w', encoding='utf-8') as file:
    json.dump(data, file, ensure_ascii=False, indent=2)
```

## Merge two jsons

```python
# Merge data and data2 into a single list
merged_data = data + data2

# Save the merged data to a new file
merged_filename = r"d:\UPSC-AI\Data\mock prelims\data\qa-merged.json"
with open(merged_filename, 'w', encoding='utf-8') as file:
    json.dump(merged_data, file, ensure_ascii=False, indent=2)

# Verify the merge (optional)
print(f"Number of items in merged data: {len(merged_data)}")
```
